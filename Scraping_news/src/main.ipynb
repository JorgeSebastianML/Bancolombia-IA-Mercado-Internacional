## Busqueda Semanal

Se realiza una busqueda semanal

```python
# Se importa la libreria de analisis
from news_analisis import newsAnalisis
# Se inicaliza la clase newsAnalisis
analizer = newsAnalisis()
# Se define la key para consumir los servicios de chatgpt
key = "API_KEY"
# Se crea una lista de busqueda (Tambien se le puede pasar el path de un archivo csv)
search_list = ["PROCTER & GAMBLE-PG-NEWS"]
# Se establece la fecha final de la busqueda
end_date = '2021-05-30'
# Se establece que periodo de tiempo se desea realizar la busqueda
duration = 2
# Se realiza el scraping y clasificacion de la noticias
pd_news_results = analizer.search(key, search_list, end_date, duration, type_time = "W")
# Se muestran los resultados
pd_news_results.head()
```

```python
import pandas as pd
total_news_results = pd.read_csv("../temp/temp_save_results.csv")
total_news_results = pd.concat([pd_news_results, total_news_results])
total_news_results.to_csv("../temp/temp_save_results.csv", index=False)
```
