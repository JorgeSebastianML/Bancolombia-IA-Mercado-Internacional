# Historico de noticias

Se crea un historico de las noticias y se cargan a la lz

```python
# Se importa la libreria de analisis, pandas y tqdm
import pandas as pd
from tqdm import tqdm
from news_analisis import newsAnalisis
# Se inicaliza la clase newsAnalisis
analizer = newsAnalisis()
# Se define la key para consumir los servicios de chatgpt
key = "API_KEY"
# Se declara el path de un archivo csv que contiene las busquedas de interes
search_list = "../include/New_searchs.xlsx"
# Carga la lista de busquedas
search_list = ["META PLATFORMS A-META-NEWS", "AMAZON.COM-AMZN-NEWS", "PROCTER & GAMBLE-PG-NEWS", "EXXON MOBIL-XOM-NEWS", 
               "BERKSHIRE HATHAWAY 'B'-BRK.A-NEWS", "JOHNSON CONTROLS INTL.-JCI-NEWS", "RAYTHEON TECHNOLOGIES-RTX-NEWS", 
               "LINDE-LIN-NEWS", "AMERICAN TOWER-AMT-NEWS", "APPLE-AAPL-NEWS", "NEXTERA ENERGY-NEE-NEWS"]
# Se establece la fecha final de la busqueda
end_date = '2021-07-01'
# Se establece que periodo de tiempo se desea realizar la busqueda
duration = 2
# Se crea un dataframe vacio 
try:
    total_news_results = pd.read_csv("../temp/temp_save_results.csv")
except:
    total_news_results = pd.DataFrame()
# Se recorre la lista de busquedas
for i in tqdm(range(len(search_list))):
    if i < 3:
        pass
    else:
        print("Procesando " + search_list[i])
        # Se realiza el scraping y clasificacion de la noticias de la busqueda puntual
        pd_news_results = analizer.search(key, [search_list[i]], end_date, duration, type_time = "MS")
        # Se agrega los resultados el dataframe total
        total_news_results = pd.concat([pd_news_results, total_news_results])
        # Se guarda un temporal de los resultados en un csv
        #total_news_results.to_csv("../temp/temp_save_results.csv", index=False)

        break
```
