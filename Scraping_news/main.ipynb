```python
from datetime import datetime
import threading
import time

DELAY_REQUESTS = 0
DELAY_RUN = 0


class Scraper:

    def __init__(self, name, company_serch, fecha, periodo, type_time, result_queue):
        self.name = name
        self.company_serch = company_serch
        self.fecha = fecha
        self.periodo = periodo
        self.type_time = type_time
        self.result = []
        self.result_queue = result_queue
        
    def exec(self):
        st = time.time()
        ht = datetime.now().time().strftime("%H:%M:%S")
        threads = [threading.Thread(target=self.run, args=(param,)) for param in self.company_serch]
        for thread in threads:
            time.sleep(DELAY_RUN)
            thread.start()

        for thread in threads:
            thread.join()

        self.result_queue.put({"name": self.name,
                               "fecha": self.fecha,
                               "ht": ht,
                               "t": time.time() - st,
                               "result": self.result})
        return self.result

    def run(self, param):
        st = time.time()
        self.result.append({"param": param,
                            "t": time.time() - st,
                            "df": scrap.GoogleNews([param], self.fecha, self.periodo, type_time = self.type_time)})
```
